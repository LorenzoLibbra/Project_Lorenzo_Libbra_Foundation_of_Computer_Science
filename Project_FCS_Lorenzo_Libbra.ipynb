1. Extract all students that study Online, have an attendance_rate smaller than 50, and have more than 5 missed_deadlines. Store them in a dataframe called at_risk.

```python
import pandas as pd
```

```python
#Displaying the students table called student
student = pd.read_csv("student_dropout_risk_2026.csv")
student
```

```python
#Displaying student's columns
student.columns
```

```python
#Creation of the new table (at_risk), obtained by filtering the student table based on the three required conditions for 'study_mode','attendance_rate' and 'missed_deadlines'.
at_risk = student[(student['study_mode'] == 'Online') & (student['attendance_rate'] < 50) & (student['missed_deadlines'] > 5)]
at_risk
```

2. Then compute, for each course_type, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called at_risk_by_course, with columns students and dropout_rate, sorted by decreasing dropout_rate.

```python
#Solution obtained using the methods covered in class
th_students = at_risk.groupby('course_type')['person_id'].count().rename('students') #Grouping on 'course_type' and counting 'person_id' later called 'students'
dropout = at_risk.groupby('course_type')['dropout'].mean().rename('dropout_rate') #Grouping on 'course_type' and executing the mean on 'dropout' later called 'dropout_rate'
at_risk_by_course = pd.merge(th_students, dropout, left_on='course_type', right_on='course_type') #Merging the th_students table and the dropout table
at_risk_by_course = at_risk_by_course.sort_values('dropout_rate', ascending=False) #Sorting the new table at_risk_by_course with a descending order
at_risk_by_course
```

```python
#A simpler and faster-to-write alternative
at_risk_by_course = at_risk.groupby('course_type').agg(students=('person_id', 'count'), dropout_rate=('dropout', 'mean'))
at_risk_by_course = at_risk_by_course.sort_values('dropout_rate', ascending=False)
at_risk_by_course
```

3. Add to the dataframe the following columns:

- gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and "stable" otherwise.

7. Build a risk score by summing:
- 2 points if the tuition is not up to date;  
- 1 point if stress_level is at least 8;
- 1 point if sleep_hours is smaller than 5;
- 1 point if the student has a part-time job with more than 20 weekly work hours;
- 2 points if attendance_rate is smaller than 60;
- 1 point if missed_deadlines is larger than 3.

For each value of the risk score, compute the number of students and the dropout rate.

```python
#Definition of the 'risk_score' function, used to sum the points associated to the six required characteristics.
def risk_score(table):
  points = 0 #score initialization
  if table['tuition_up_to_date'] == 0: #2 points if the tuition is not up to date;
    points += 2
  if table['stress_level'] >= 8: #1 point if stress_level is at least 8;
    points += 1
  if table['sleep_hours'] < 5: #1 point if sleep_hours is smaller than 5;
    points += 1
  if (table['part_time_job'] == 1) & (table['weekly_work_hours'] > 20): #1 point if the student has a part-time job with more than 20 weekly work hours;
    points += 1
  if table['attendance_rate'] < 60: #2 points if attendance_rate is smaller than 60;
    points += 2
  if table['missed_deadlines'] > 3: #1 point if missed_deadlines is larger than 3.
    points += 1
  return points #score result

student['risk_score'] = student.apply(risk_score, axis=1) #creation of a new column in 'student' called 'risk_score'
student
```

```python
#Computiing the number of students and the dropout rate
n_students = student.groupby('risk_score')['person_id'].count().rename('number_of_students') #Grouping on 'risk_score' and counting 'person_id' later called 'number_of_students'
drop = student.groupby('risk_score')['dropout'].mean().rename('dropout_rate') #Grouping on 'risk_score' and executing the mean on 'dropout' later called 'dropout_rate'
risk_score_numbers = pd.merge(n_students, drop, left_on='risk_score', right_on='risk_score') #Joining the n_students table and the drop table
risk_score_numbers[['number_of_students','dropout_rate']] #Selecting only the columns of interest
#Similarly to the second exercise, the agg() method could have been used.
```

8. Read the GDP per capita dataset published by Our World in Data directly from the URL https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true and store it in a dataframe called gdp.\
The column ny_gdp_pcap_pp_kd contains the GDP per capita. \
Clean the dataset, removing the rows that do not correspond to a single country (e.g. World, High-income countries). For each country keep only the row with the most recent year.

```python
#Displaying the gdp table
gdp = pd.read_csv('gdp-per-capita-worldbank.csv')
gdp
```

```python
gdp = gdp[gdp['owid_region'].notnull()] #Filtering the column 'ny_gdp_pcap_pp_kd' to search everything different from '0' and where 'owid_region' is not null
gdp = gdp.loc[gdp.groupby('entity')['year'].idxmax()] #Filtering 'year' to search the most recent year
gdp
```

9. Join the GPD data with the students data. Compute the dropout rate for each continent.

```python
#Merging the gdp table and the student table, considering all the continents with how='left'
gdp_students = pd.merge(gdp, student, left_on='entity', right_on='nationality', how='left')
continent_rate = gdp_students.groupby('owid_region')['dropout'].mean().sort_values().rename('dropout_rate') #Grouping by 'owid_region' and taking the 'dropout_rate' in order.
continent_rate
```
